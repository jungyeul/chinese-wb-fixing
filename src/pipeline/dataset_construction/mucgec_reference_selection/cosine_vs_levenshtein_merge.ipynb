```python
import pandas as pd

# Load the two CSV files
cos_df = pd.read_csv('../data/cosine_target.csv')
lev_df = pd.read_csv('../data/levenshtein_target.csv')

# Rename the target column for consistency
cos_df = cos_df.rename(columns={'target': 'cosine_target'})
lev_df = lev_df.rename(columns={'target': 'levenshtein_target'})

# Merge the two dataframes on 'id' and 'source'
merged = pd.merge(cos_df, lev_df, on=['id', 'source'], how='inner')

# Add a comparison column to indicate whether the targets are the same
merged['is_same'] = merged['cosine_target'] == merged['levenshtein_target']

# (Optional) Replace True/False with checkmark and cross symbols
# merged['is_same'] = merged['is_same'].map({True: '✅', False: '❌'})

# Save the merged result to a new CSV file
merged.to_csv('cosine_vs_levenshtein.csv', index=False, encoding='utf-8-sig')

print("✅ Merge complete. Saved as cosine_vs_levenshtein.csv")
```
